User Behaviour Analysis

```python
# Calculate unique users for each event type

user_event_summary = (
    events.groupby("event")
    .agg(
        total_events=("visitorid", "size"),
        unique_users=("visitorid", "nunique")
    )
    .reset_index()
)

display(user_event_summary)
```

```python
# Create sets of users for each event

view_users = set(
    events.loc[events["event"] == "view", "visitorid"]
)

cart_users = set(
    events.loc[events["event"] == "addtocart", "visitorid"]
)

purchase_users = set(
    events.loc[events["event"] == "transaction", "visitorid"]
)

# Calculate user overlap

user_overlap = {
    "View Users": len(view_users),
    "Cart Users": len(cart_users),
    "Purchase Users": len(purchase_users),

    "View and Cart": len(view_users & cart_users),

    "View and Purchase": len(view_users & purchase_users),

    "Cart and Purchase": len(cart_users & purchase_users),

    "All Three Events": len(
        view_users & cart_users & purchase_users
    )
}

display(
    pd.DataFrame(
        user_overlap.items(),
        columns=["User Category", "Unique Users"]
    )
)
```

```python
# Users who viewed but never added to cart or purchased

view_only_users = (
    view_users - cart_users - purchase_users
)

print(
    "Users who only viewed products:",
    len(view_only_users)
)

print(
    "Percentage of viewing users:",
    round(
        len(view_only_users) / len(view_users) * 100,
        2
    )
)
```

```python
# Preliminary user-level conversion metrics

view_to_cart = (
    len(view_users & cart_users)
    / len(view_users) * 100
)

view_to_purchase = (
    len(view_users & purchase_users)
    / len(view_users) * 100
)

cart_to_purchase = (
    len(cart_users & purchase_users)
    / len(cart_users) * 100
)

print("View to Cart:", round(view_to_cart, 2), "%")

print("View to Purchase:", round(view_to_purchase, 2), "%")

print("Cart to Purchase:", round(cart_to_purchase, 2), "%")
```

quantify chronological product-level conversion

```python
# 1. Find the first recorded view for each visitor-product pair

views = (
    events[events["event"] == "view"]
    .groupby(["visitorid", "itemid"])["timestamp"]
    .min()
    .reset_index(name="first_view")
)

# 2. Find cart events occurring after the first view

carts = events.loc[
    events["event"] == "addtocart",
    ["visitorid", "itemid", "timestamp"]
]

cart_after_view = carts.merge(
    views,
    on=["visitorid", "itemid"],
    how="inner"
)

cart_after_view = cart_after_view[
    cart_after_view["timestamp"] > cart_after_view["first_view"]
]

# Keep the first qualifying cart event for each visitor-product pair
cart_after_view = (
    cart_after_view
    .groupby(["visitorid", "itemid"])["timestamp"]
    .min()
    .reset_index(name="first_cart_after_view")
)

# 3. Find transactions occurring after a qualifying cart event

transactions = events.loc[
    events["event"] == "transaction",
    ["visitorid", "itemid", "timestamp"]
]

purchase_after_cart = transactions.merge(
    cart_after_view,
    on=["visitorid", "itemid"],
    how="inner"
)

purchase_after_cart = purchase_after_cart[
    purchase_after_cart["timestamp"] >
    purchase_after_cart["first_cart_after_view"]
]

purchase_after_cart = (
    purchase_after_cart
    .groupby(["visitorid", "itemid"])["timestamp"]
    .min()
    .reset_index(name="first_purchase_after_cart")
)

# 4. Summarise the sequential product funnel

funnel_summary = pd.DataFrame({
    "Stage": [
        "Viewed Product",
        "Viewed Then Added to Cart",
        "Viewed, Carted Then Purchased"
    ],
    "Unique Visitor-Product Pairs": [
        len(views),
        len(cart_after_view),
        len(purchase_after_cart)
    ],
    "Unique Visitors": [
        views["visitorid"].nunique(),
        cart_after_view["visitorid"].nunique(),
        purchase_after_cart["visitorid"].nunique()
    ]
})

display(funnel_summary)

# 5. Calculate sequential conversion rates

view_to_cart_rate = (
    len(cart_after_view) / len(views) * 100
)

cart_to_purchase_rate = (
    len(purchase_after_cart) / len(cart_after_view) * 100
)

view_to_purchase_rate = (
    len(purchase_after_cart) / len(views) * 100
)

print("Sequential View to Cart:", round(view_to_cart_rate, 2), "%")
print("Sequential Cart to Purchase:", round(cart_to_purchase_rate, 2), "%")
print("Sequential View to Purchase:", round(view_to_purchase_rate, 2), "%")
```

```python
# Reconstruct the time between the first view and qualifying cart event
view_to_cart_timing = views.merge(
    cart_after_view,
    on=["visitorid", "itemid"],
    how="inner"
)

view_to_cart_timing["view_to_cart_hours"] = (
    view_to_cart_timing["first_cart_after_view"]
    - view_to_cart_timing["first_view"]
) / (1000 * 60 * 60)


# Reconstruct the time between qualifying cart and purchase
purchase_timing = cart_after_view.merge(
    purchase_after_cart,
    on=["visitorid", "itemid"],
    how="inner"
)

purchase_timing["cart_to_purchase_hours"] = (
    purchase_timing["first_purchase_after_cart"]
    - purchase_timing["first_cart_after_view"]
) / (1000 * 60 * 60)


# Summarise time gaps
print("VIEW TO CART TIME GAP (HOURS)")
display(
    view_to_cart_timing["view_to_cart_hours"]
    .describe(percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99])
)

print("CART TO PURCHASE TIME GAP (HOURS)")
display(
    purchase_timing["cart_to_purchase_hours"]
    .describe(percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99])
)
```

Conversion-Window Sensitivity Analysis

```python
# Combine view-to-cart and cart-to-purchase timing
full_journey_timing = view_to_cart_timing.merge(
    purchase_timing,
    on=["visitorid", "itemid"],
    how="inner"
)

# Candidate conversion windows
windows = {
    "1 Hour": 1,
    "24 Hours": 24,
    "7 Days": 168
}

window_results = []

for window_name, hours in windows.items():

    # View-to-cart conversion within the window
    view_cart_count = (
        view_to_cart_timing["view_to_cart_hours"] <= hours
    ).sum()

    # Cart-to-purchase conversion within the window
    cart_purchase_count = (
        purchase_timing["cart_to_purchase_hours"] <= hours
    ).sum()

    # Complete journey: both transitions within the window
    complete_count = (
        (
            full_journey_timing["view_to_cart_hours"] <= hours
        ) &
        (
            full_journey_timing["cart_to_purchase_hours"] <= hours
        )
    ).sum()

    window_results.append({
        "Conversion Window": window_name,
        "View to Cart Pairs": view_cart_count,
        "View to Cart Rate (%)": round(
            view_cart_count / len(views) * 100, 2
        ),
        "Cart to Purchase Pairs": cart_purchase_count,
        "Cart to Purchase Rate (%)": round(
            cart_purchase_count / len(cart_after_view) * 100, 2
        ),
        "Complete Funnel Pairs": complete_count,
        "View to Purchase Rate (%)": round(
            complete_count / len(views) * 100, 2
        )
    })

window_analysis = pd.DataFrame(window_results)

display(window_analysis)
```

Calculating end-to-end conversion time

```python
# Calculate total time from first view to subsequent purchase
full_journey_timing["total_conversion_hours"] = (
    full_journey_timing["first_purchase_after_cart"]
    - full_journey_timing["first_view"]
) / (1000 * 60 * 60)

# Compare strict end-to-end conversion windows
end_to_end_results = []

for window_name, hours in windows.items():

    completed_pairs = (
        full_journey_timing["total_conversion_hours"] <= hours
    ).sum()

    end_to_end_results.append({
        "Conversion Window": window_name,
        "Complete Funnel Pairs": completed_pairs,
        "View to Purchase Rate (%)": round(
            completed_pairs / len(views) * 100, 2
        )
    })

end_to_end_analysis = pd.DataFrame(end_to_end_results)

display(end_to_end_analysis)

# Inspect the total conversion time distribution
display(
    full_journey_timing["total_conversion_hours"]
    .describe(
        percentiles=[0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
    )
)
```
